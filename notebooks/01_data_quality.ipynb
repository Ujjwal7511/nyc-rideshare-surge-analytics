# 01 Data quality
Reads `reports/cleaning_log.md` and warehouse counts. Run the pipeline first.

```python
from pathlib import Path
import duckdb
root = Path('..') if Path('src').exists() is False else Path('.')
if not (root / 'src').exists():
    root = Path('.')
print((root / 'reports' / 'cleaning_log.md').read_text()[:2000])
con = duckdb.connect(str(root / 'data' / 'processed' / 'warehouse.duckdb'), read_only=True)
print(con.execute('SELECT COUNT(*) raw FROM raw_trips').fetchdf())
print(con.execute('SELECT COUNT(*) clean FROM trips_clean').fetchdf())
print(con.execute('SELECT MIN(pickup_datetime), MAX(pickup_datetime) FROM trips_clean').fetchdf())
```
